**Story**

Imagine you work as a Data Scientist in an insurance company.

The company has 1300 previous insurance claim cases.

Whenever a person meets with an accident, they file an insurance claim.

Some customers:

✅ Hire an attorney (lawyer)

Some customers:

❌ Don't hire an attorney.

**Why does the company care?**

Suppose two customers meet with the same accident.

**Customer A**
Doesn't hire an attorney.
Insurance company pays ₹2 lakh.
    
**Customer B**
Hires an attorney.
Attorney negotiates.
Insurance company pays ₹5 lakh.

Notice what happened?

The insurance company paid ₹3 lakh more.

This increases the company's cost.

**Business Problem**

The insurance company wants to know:

Before the claim process starts, can we predict whether a customer will hire an attorney?

If they know this in advance, they can prepare better.

**Why is this useful?**

Suppose the model predicts:

Customer A

Probability of hiring attorney = 95%

The company can:

Prepare legal documents.
Assign an experienced claim officer.
Estimate the claim amount.

Instead of being surprised later.

**Business Objective**

The company wants to answer two questions.

**Question 1**

**Can we predict whether a customer will hire an attorney?**

Example:

Customer

↓

Prediction

↓

Yes

**Question 2**

**Which type of customers usually hire attorneys?**

Example:

Maybe customers with:

Severe injuries
Higher medical expenses
Longer hospital stays

hire attorneys more often.

These insights help the company improve its business decisions.

**Business Benefit**

If the company knows this early,

it can:

- Estimate future claim costs.
- Prepare legal teams.
- Reduce delays.
- Improve customer service.
- Increase profitability.
