### LINEAR REGRESSION: [REGRESSION PROBLEM]

* Linear regression is a supervised learning algorithm used to predict continuous numerical values.
  
* It learns the relationship between an input (X) and output variables (y) by fitting the best fit line and predict values of new data.

* **Linear regrrsssion line or best fitted line or regression line**

1. SIMPLE LINEAR REGRESSION:

* If we are predicting output based on one input feature .
* salary based on experience

$$y=MX+C$$

2.  MULTIPLE LINEAR REGRESSION:

* If we are predicting output based on more than one input feature, that is multiple linear regression.
* Salary based on experience, age, skills.

$$y = M_1X_1 + M_2X_2 + M_3X_3 + C$$

**What does "Linear" mean?**

Linear means:

The relationship between the input (X) and output (Y) can be represented by a straight line.

### why best-fitted line ?

* Most of the times real data is messy.
* So, line cannot pass through every point.
* Machine tries to draw one line that is closest to all the points.
* * **Linear regrrsssion line or best fitted line or regression line**
 
* The Best Fit Line is the line that best represents the relationship between the independent and dependent variables. It minimizes the residual errors, allowing the model to make the most accurate predictions possible.

**"Does Linear Regression always draw a straight line?"**


In Simple Linear Regression, the model fits a straight line because there is one independent variable. In Multiple Linear Regression, the prediction is still linear in the coefficients, but the model is represented as a plane or hyperplane instead of a line because it uses multiple independent variables.

**Why do we check assumptions?**

We check the assumptions of Linear Regression to ensure that the model is valid and produces reliable predictions. Linear Regression is based on statistical assumptions such as linearity, independence of errors, homoscedasticity, normality of residuals, and no multicollinearity. If these assumptions are violated, the model's predictions and interpretations may become unreliable.
