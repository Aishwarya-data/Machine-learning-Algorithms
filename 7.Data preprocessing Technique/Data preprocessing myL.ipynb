#### 2.One hotencoder

This can be achieved by using any of the following 2 libraries:

- Pandas - pd.get_dummies()
- sklearn - OneHotEncoder()

One-Hot Encoding is a feature engineering technique used to convert categorical variables into binary (0 and 1) columns. Each unique category becomes a separate column, where 1 indicates the presence of that category and 0 indicates its absence. It is commonly used for nominal categorical data in machine learning.

##### 1.Pandas - pd.get_dummies()

Convert categorical variable into dummy/indicator variables.

Each variable is converted in as many 0/1 variables as there are different
values. Columns in the output are each named after a value; if the input is
a DataFrame, the name of the original variable is prepended to the value.


```python
weather_datacopy1 = weather_data.copy()
weather_datacopy1
```

```python
weather_datacopy1 = pd.get_dummies(weather_datacopy1,columns=["Weather"],dtype=int)
weather_datacopy1
```

##### 2.sklearn - OneHotEncoder()

```python
weather_datacopy2 = weather_data.copy()
weather_datacopy2
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output=False)
encoded= ohe.fit_transform(weather_datacopy2[["Weather"]])
encoded
```

```python
encoded_weather_datacopy2 = pd.DataFrame(encoded,columns=ohe.get_feature_names_out(['Weather']))
encoded_weather_datacopy2
```

```python
weather_datacopy2 = weather_datacopy2.drop('Weather', axis=1)

weather_datacopy2 = pd.concat(
    [weather_datacopy2, encoded_weather_datacopy2],
    axis=1
)

weather_datacopy2.head()
```

```python
#del weather_datacopy2
```

**Why can't we store the encoded output in a single Weather column?**

Because the Weather column has three unique categories (C, PS, and S). **OneHotEncoder creates one binary column for each unique category.** A value of 1 indicates the presence of that category, while 0 indicates its absence.

### How to choose between label encoder and  one hot encoder ?

### Input feature:
Based on whether it is **Parametric or a Non Parametric Model**, we will choosing between LE or OHE.
* For **Parametric Models**[Linear or logistic reg] - better to go with **OHE.**
* For **Non-Parametric Models** [DT]- always go with **Label Encoding.**

### Output feature:
Always the output feature, it must be **Label Encoder.**
