## 5.Model Building

### 5.1 splitting input outputs and Feature scaling

```python
X = wine_data_df.drop("target",axis = 1)
y = wine_data_df["target"]
```

```python
std_scaler = StandardScaler()
scaled_X = std_scaler.fit_transform(X)
```

### 5.2train test split

```python
X_train,X_test,y_train,y_test = train_test_split(scaled_X,y,test_size=0.20,random_state=11,shuffle=True,stratify=y)
```

### How to choose optimal no of neighbours ?   -- using for loop

```python
from sklearn.model_selection import KFold,cross_val_score
```

#### Create K-Fold Cross Validation

```python
kfold = KFold(n_splits=5, shuffle=True, random_state=123)
cv_scores_mean=[]
```

```python
for i in range(3,52,2):
    knn_classifier = KNeighborsClassifier(n_neighbors=i)
    knn_classifier.fit(scaled_X,y)
    cv_scores = cross_val_score(estimator=knn_classifier , X=scaled_X,y=y, cv =kfold)
    cv_scores_mean.append(cv_scores.mean())
```

```python
best_score = max(cv_scores_mean)

print(best_score)
```

```python
best_k = range(3,52,2)[cv_scores_mean.index(best_score)]

print(best_k)
```

### Visualization

```python
from matplotlib import pyplot as plt

plt.plot(range(3,52,2),cv_scores_mean)
plt.xlabel("No.of Neighbours")
plt.ylabel("Mean CV Score")
plt.show()
```

**Why do we use K-Fold Cross Validation to find the best K?**


Instead of evaluating KNN on a single train-test split, we use K-Fold Cross Validation to test different values of K on multiple train-test splits. We then select the value of K that gives the highest average cross-validation score, making the model selection more reliable and less dependent on a particular split of the data
