## 5.Model Building

### 5.1 splitting input and output feature

```python
X = breast_cancer_data_df.drop(labels = "target",axis=1)
```

```python
y= breast_cancer_data_df["target"]
```

```python
X.shape,y.shape
```

```python
y.value_counts()
```

```python
from matplotlib import pyplot as plt
plt.pie(x=y.value_counts() ,labels= ["Benign","Malignant"],autopct="%1.0f%%")
plt.show()
```

### 5.2 Data preprocessing

#### Train test split

```python
from sklearn.model_selection import train_test_split
```

##### Before using stratify 

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,shuffle=True,random_state=11)
X_train,X_test,y_train,y_test
```

```python
print("Training samples: " , X_train,y_train)
print("Testing samples: ", X_test,y_test)
```

```python
plt.pie(y_train.value_counts(),labels=["Benign","Malignant"],autopct="%1.0f%%")
plt.show()
```

```python
plt.pie(y_test.value_counts(),labels=["Benign","Malignant"],autopct="%1.0f%%")
plt.show()
```

##### After stratify sampling

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,shuffle=True,random_state=11,stratify=y)
X_train,X_test,y_train,y_test
```

```python
plt.pie(y_train.value_counts(),labels=["Benign","Malignant"],autopct="%1.0f%%")
plt.show()
```

```python
plt.pie(y_test.value_counts(),labels=["Benign","Malignant"],autopct="%1.0f%%")
plt.show()
```

#### When should we use stratify sampling?

We should use stratify in classification problems, especially when the target classes are imbalanced, to ensure that the train and test sets have the same class proportions as the original dataset.

**stratify = Preserve the class ratio.**



## 6.Model Training

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
%%time
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=3,class_weight={0:2,1:1})
dt_model.fit(X_train,y_train)
```

**class_weight** is used to assign different importance to different classes during the training of a Decision Tree. It is mainly used for imbalanced datasets, where the minority class receives a higher weight. This encourages the model to pay more attention to minority-class samples and reduces bias toward the majority class.

Here class_weight={0:2,1:1}   0 --> Malinant , 1--> beniGn  , malignant is dangerous so giving more importance to  0 

## 7.Model Testing & 8.Model Evaluation

### Training data

```python
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

### Evaluation metrics

```python
from sklearn.metrics import classification_report ,confusion_matrix
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

### Testing data

```python
y_pred_test = dt_model.predict(X_test)
y_pred_test
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```
